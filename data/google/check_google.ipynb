```python
import pickle as pkl 
import pandas as pd 
import numpy as np 
import json 

def read_pkl(path):
    with open(path, 'rb') as f:
        return pkl.load(f)

def slash():
    print('-' * 30)

def equal_str_(a, b):
    a = a.split()
    b = b.split()
    if len(a) != len(b):
        return False
    for i,j in zip(a, b):
        if i != j:
            return False
    return True
```
